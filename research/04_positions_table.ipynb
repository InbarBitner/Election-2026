```python
import glob
import pandas as pd

files = glob.glob('platform_coverage_*.csv')
print(files)

positions = pd.concat([pd.read_csv(f) for f in files], ignore_index=True)
positions = positions[['party', 'question_id', 'axis', 'stance', 'confidence', 'quote']]
print(positions.shape)
```

```python
table = positions.pivot(index='question_id', columns='party', values='stance')
table
```

```python
print(table.notna().sum(axis=1))
```

```python
print(table.notna().sum(axis=0))
```

```python
has_for = (table == 1).any(axis=1)
has_against = (table == -1).any(axis=1)
divisive = has_for & has_against

table[divisive]
```
